# Data curation

The raw pages remain unchanged. This step keeps all incident rows, preserves missing values, sets dispatch duration to missing only when FDNY marks its components invalid, and adds clear quality flags. It does not remove statistical outliers or create a long-response target.

```python
from pathlib import Path
import json, subprocess, sys
ROOT = Path.cwd()
if not (ROOT / 'etl' / 'curate_dataset.py').exists():
    ROOT = ROOT.parent
curated_file = ROOT / 'data' / 'curated' / 'ems_incident_dispatch_curated_2024-01_to_2026-06.csv.gz'
summary_file = ROOT / 'reports' / 'curation_summary.json'
if not curated_file.exists():
    subprocess.run([sys.executable, str(ROOT / 'etl' / 'curate_dataset.py')], cwd=ROOT, check=True)
else:
    print('Using the saved curated file. Move it aside before creating a fresh copy.')
summary = json.loads((ROOT / 'reports' / 'curation_summary.json').read_text(encoding='utf-8'))
summary
```

# Exploratory data analysis

This notebook reads the saved profile outputs. Response-time averages use the FDNY validity flag (`Y`). Counts are descriptive totals and are not adjusted for population or staffing. See `reports/eda_report.md` for the written findings.

```python
from pathlib import Path
import pandas as pd
from IPython.display import Image, display
ROOT = Path.cwd()
if not (ROOT / 'reports' / 'raw_profile_summary.json').exists():
    ROOT = ROOT.parent
REPORTS = ROOT / 'reports'
```

```python
display(pd.read_csv(REPORTS / 'incidents_by_month.csv'))
display(pd.read_csv(REPORTS / 'incidents_by_weekday.csv'))
display(pd.read_csv(REPORTS / 'incidents_by_hour.csv'))
```

```python
display(pd.read_csv(REPORTS / 'frequencies_borough.csv').head(10))
display(pd.read_csv(REPORTS / 'frequencies_initial_call_type.csv').head(15))
display(pd.read_csv(REPORTS / 'frequencies_final_call_type.csv').head(15))
display(pd.read_csv(REPORTS / 'severity_transitions.csv'))
```

```python
display(pd.read_csv(REPORTS / 'response_by_borough.csv').sort_values('average_response_seconds', ascending=False))
display(pd.read_csv(REPORTS / 'response_by_initial_call_type.csv').sort_values('valid_response_count', ascending=False).head(20))
display(pd.read_csv(REPORTS / 'unmapped_documented_codes.csv'))
```
